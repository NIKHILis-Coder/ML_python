## 2. Opening a CSV file from a URL

```python
import requests
from io import StringIO

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## 12. handling dates

```python
import random
from datetime import datetime, timedelta

df = pd.read_csv('titanic.csv')
base = datetime(1912, 4, 10)
df['boarding_date'] = [base + timedelta(hours=random.randint(0, 48)) for _ in range(len(df))]
df.to_csv('titanic_with_dates.csv', index=False)
df[['Name', 'boarding_date']].head()
```

```python
df = pd.read_csv('titanic_with_dates.csv')
print(df['boarding_date'].dtype)
```

```python
df = pd.read_csv('titanic_with_dates.csv', parse_dates=['boarding_date'])
print(df['boarding_date'].dtype)
df[['Name', 'boarding_date']].head()
```

```python
df['day'] = df['boarding_date'].dt.day
df['hour'] = df['boarding_date'].dt.hour
df[['Name', 'boarding_date', 'day', 'hour']].head(8)
```
