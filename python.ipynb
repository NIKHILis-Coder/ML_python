```python
y = lambda x : x+2
print(y(4))

list1 = [1,2,3,4,5,6]
res = list(map(y, list1))
print(res)

def square(n):
    return n*n

res = list(map(square, list1))
print(res)
```

```python
left = pd.DataFrame({
    'key': ['K0', 'K1', 'K2', 'K3'],
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B': ['B0', 'B1', 'B2', 'B3']
})

right = pd.DataFrame({
    'key': ['K0', 'K1', 'K2', 'K3'],
    'C': ['C0', 'C1', 'C2', 'C3'],
    'D': ['D0', 'D1', 'D2', 'D3']
})
```

```python
pd.merge(left, right, how ='outer', on =['key'])
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linespace(1,15,9)
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2,
                         figsize=(8,8),
                         dpi=100)

axes[0,0].plot(x, x**2, color ='r', label='xsquare')
axes[0,0].legend()

axes[0,1].plot(x, x**3, 'g', label='xcube')
axes[0,1].legend()

axes[1,0].plot(x, x**4, 'b', label='xraiseto4')
axes[1,0].legend()

axes[1,1].plot(x, x**7, 'r--', label='nothing')
axes[1,1].legend()

fig.suptitle('Practice')

plt.tight_layout()

plt.show()
```
