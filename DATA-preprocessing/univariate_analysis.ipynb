```python
Q1 = df['body_weight'].quantile(0.25)
Q3 = df['body_weight'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

filtered_df = df[
    (df['body_weight'] >= lower) &
    (df['body_weight'] <= upper)
]
```
