### WORKING WITH DATES

```python
date['date']= pd.to_datetime(date['date'])
date['date'].dtype
date.info()
```

### extract year

```python
date['date_year']= date['date'].dt.year
date['date_year']
```

#### EXTRACT MONTH


```python
date['date_month_no']= date['date'].dt.month
date.head()
```

### EXTRACT MONTH NAME

```python
date['date_month_name']= date['date'].dt.month_name()
date.head()
```

### EXTACT DAYS

```python
date['date_day']= date['date'].dt.day
date.head()
```

### day of week

```python
date['date_dow']= date['date'].dt.dayofweek
date.head()
```

### day of week- name

```python
date['date_dow_name']= date['date'].dt.day_name()
date.head()
```

### Is Weekend?

```python
date['date_is_weekend'] = np.where(
    date['date_dow_name'].isin(['Sunday', 'Saturday']),
    1,
    0
)
date.head()
```

### Extract Week of Year

```python
date['date_week'] = date['date'].dt.isocalendar().week
date.head()
```

Step 1: Create Quarter

```python
date['quarter'] = date['date'].dt.quarter
```

### 3. Extract Semester

```python
date['semester'] = np.where(
    date['quarter'].isin([1, 2]),
    1,
    2
)
date.head()
```

```python
import datetime

# Get current date and time
today = datetime.datetime.today()

# Display current date and time
print(today)
```

```python
# Difference between today and the date column
# Result is a Timedelta object

date['days_passed'] = (today - date['date']).dt.days

# View result
date[['date', 'days_passed']].head()
```

```python
import datetime

today = datetime.datetime.today()

date['months_passed'] = (
    (today.year - date['date'].dt.year) * 12
    + (today.month - date['date'].dt.month)
)

date[['date', 'months_passed']].head()
```
