```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'linear_regression\placement (1).csv')
df
```

```python
X= df.iloc[:,0:1]
y= df.iloc[:,1:]
```

```python
class GRADEINT_DESCENT:

    def __init__(self, learning_rate, epochs, int_m, int_b):
        self.m = int_m
        self.b = int_b
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X_train, y_train):

        n = len(X_train)

        for i in range(self.epochs):

            y_pred = self.m * X_train.ravel() + self.b

            slope_gd = (-2/n) * np.sum((y_train - y_pred) * X_train)

            intercept_gd = (-2/n) * np.sum(y_train - y_pred)

            self.m = self.m - self.lr * slope_gd
            self.b = self.b - self.lr * intercept_gd

        print(self.m, self.b)
```
